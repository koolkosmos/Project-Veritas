```python
import pandas as pd
import os

# Load CSVs
fake_df = pd.read_csv(f"{CSV_DIR}/Deepfakes.csv")
real_df = pd.read_csv(f"{CSV_DIR}/original.csv")

# Build paths
fake_df["video_path"] = fake_df["File Path"].apply(
    lambda x: os.path.join(FAKE_DIR, os.path.basename(x))
)
real_df["video_path"] = real_df["File Path"].apply(
    lambda x: os.path.join(REAL_DIR, os.path.basename(x))
)

# Assign labels
fake_df["label"] = 1
real_df["label"] = 0

# Keep only required columns
fake_df = fake_df[["video_path", "label"]]
real_df = real_df[["video_path", "label"]]

# Combine
df = pd.concat([fake_df, real_df], ignore_index=True)

# Sanity check
missing = df[~df["video_path"].apply(os.path.exists)]
print(f"Missing files: {len(missing)}")

df.head()
```

```python
import cv2
import numpy as np

def sample_frames_uniform(video_path, num_frames=32):
    cap = cv2.VideoCapture(video_path)
    if not cap.isOpened():
        raise ValueError(f"Cannot open video: {video_path}")

    total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
    if total_frames < num_frames:
        raise ValueError(
            f"Video too short: {total_frames} frames < {num_frames}"
        )

    # Uniformly spaced frame indices
    frame_indices = np.linspace(
        0, total_frames - 1, num_frames, dtype=int
    )

    frames = []
    current_idx = 0
    target_ptr = 0

    while cap.isOpened() and target_ptr < num_frames:
        ret, frame = cap.read()
        if not ret:
            break

        if current_idx == frame_indices[target_ptr]:
            frames.append(frame)
            target_ptr += 1

        current_idx += 1

    cap.release()

    frames = np.array(frames)

    if frames.shape[0] != num_frames:
        raise RuntimeError(
            f"Expected {num_frames} frames, got {frames.shape[0]}"
        )

    return frames
```

```python
import cv2
import numpy as np

def sample_and_preprocess_frames(
    video_path,
    num_frames=32,
    target_size=(160,160),
    normalize=True
):
    cap = cv2.VideoCapture(video_path)
    if not cap.isOpened():
        raise ValueError(f"Cannot open video: {video_path}")

    total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
    if total_frames < num_frames:
        raise ValueError(
            f"Video too short: {total_frames} frames < {num_frames}"
        )

    frame_indices = np.linspace(
        0, total_frames - 1, num_frames, dtype=int
    )

    frames = []
    current_idx = 0
    target_ptr = 0

    while cap.isOpened() and target_ptr < num_frames:
        ret, frame = cap.read()
        if not ret:
            break

        if current_idx == frame_indices[target_ptr]:
            # Resize
            frame = cv2.resize(frame, target_size)
            frames.append(frame)
            target_ptr += 1

        current_idx += 1

    cap.release()

    frames = np.array(frames, dtype=np.float32)

    if frames.shape[0] != num_frames:
        raise RuntimeError(
            f"Expected {num_frames} frames, got {frames.shape[0]}"
        )

    if normalize:
        frames /= 255.0  # [0, 1]

    return frames
```

```python
import numpy as np
import pandas as pd
import cv2
from tensorflow.keras.utils import Sequence

class VideoSequence(Sequence):
    def __init__(
        self,
        csv_path,
        batch_size=2,
        num_frames=32,
        target_size=(160, 160),
        shuffle=True
    ):
        self.df = pd.read_csv(csv_path)
        self.batch_size = batch_size
        self.num_frames = num_frames
        self.target_size = target_size
        self.shuffle = shuffle
        self.indices = np.arange(len(self.df))
        self.on_epoch_end()

    def __len__(self):
        return int(np.ceil(len(self.df) / self.batch_size))

    def on_epoch_end(self):
        if self.shuffle:
            np.random.shuffle(self.indices)

    def _sample_frames(self, video_path):
        cap = cv2.VideoCapture(video_path)
        if not cap.isOpened():
            raise ValueError(f"Cannot open video: {video_path}")

        total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
        if total_frames < self.num_frames:
            cap.release()
            raise ValueError(
                f"Video too short: {total_frames} < {self.num_frames}"
            )

        frame_indices = np.linspace(
            0, total_frames - 1, self.num_frames, dtype=int
        )

        frames = []
        current_idx = 0
        target_ptr = 0

        while cap.isOpened() and target_ptr < self.num_frames:
            ret, frame = cap.read()
            if not ret:
                break

            if current_idx == frame_indices[target_ptr]:
                frame = cv2.resize(frame, self.target_size)
                frame = frame.astype(np.float32) / 255.0
                frames.append(frame)
                target_ptr += 1

            current_idx += 1

        cap.release()

        frames = np.array(frames, dtype=np.float32)

        if frames.shape[0] != self.num_frames:
            raise RuntimeError(
                f"Expected {self.num_frames} frames, got {frames.shape[0]}"
            )

        return frames

    def __getitem__(self, idx):
        batch_indices = self.indices[
            idx * self.batch_size : (idx + 1) * self.batch_size
        ]

        batch_videos = []
        batch_labels = []

        for i in batch_indices:
            row = self.df.iloc[i]
            frames = self._sample_frames(row["video_path"])
            batch_videos.append(frames)
            batch_labels.append(row["label"])

        X = np.stack(batch_videos, axis=0)
        y = np.array(batch_labels, dtype=np.float32)

        return X, y
```

```python
import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.applications import EfficientNetB0

def build_veritas_model(
    num_frames=32,
    img_size=160,
    cnn_trainable=False,
    lstm_units=128
):
    # Input: (T, H, W, C)
    video_input = layers.Input(
        shape=(num_frames, img_size, img_size, 3),
        name="video_input"
    )

    # CNN backbone (frame-level)
    cnn = EfficientNetB0(
        include_top=False,
        weights="imagenet",
        pooling="avg"
    )
    cnn.trainable = cnn_trainable  # freeze initially

    # Apply CNN to each frame
    x = layers.TimeDistributed(cnn, name="frame_feature_extractor")(video_input)
    # Shape now: (batch, 50, feature_dim)

    # Temporal modeling
    x = layers.LSTM(lstm_units, return_sequences=False)(x)

    # Regularization
    x = layers.Dropout(0.5)(x)

    # Output: video-level prediction
    output = layers.Dense(1, activation="sigmoid", name="fake_probability")(x)

    model = models.Model(inputs=video_input, outputs=output)
    return model
```

```python
model = build_veritas_model(
    num_frames=32,
    img_size=160,
    cnn_trainable=False,  # important
    lstm_units=128
)

model.summary()
```

```python
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-4),
    loss="binary_crossentropy",
    metrics=[
        tf.keras.metrics.BinaryAccuracy(name="accuracy"),
        tf.keras.metrics.AUC(name="auc")
    ]
)
```
